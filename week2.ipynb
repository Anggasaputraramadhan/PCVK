```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
import os

file_path = 'ktp.jpg'

if not os.path.exists(file_path):
    print(f"Error: File '{file_path}' tidak ditemukan.")
    print("Silakan upload file 'ktp.jpg' terlebih dahulu ke panel Files (ikon folder di sebelah kiri Google Colab).")
else:
    img = cv.imread(file_path)
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

    h, w, _ = img_rgb.shape
    img_rb = img_rgb.copy()
    img_rb[:, :, 1] = 0

    img_gb = img_rgb.copy()
    img_gb[:, :, 0] = 0

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rb)
    plt.title("Kanal Merah - Biru (Green = 0)")

    plt.subplot(1, 2, 2)
    plt.imshow(img_gb)
    plt.title("Kanal Hijau - Biru (Red = 0)")
    plt.show()

    flip_v = cv.flip(img_rgb, 0)
    flip_h = cv.flip(img_rgb, 1)
    flip_vh = cv.flip(img_rgb, -1)

    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1)
    plt.imshow(flip_v)
    plt.title("Pencerminan Vertikal (Flip Code = 0)")

    plt.subplot(1, 3, 2)
    plt.imshow(flip_h)
    plt.title("Pencerminan Horizontal (Flip Code = 1)")

    plt.subplot(1, 3, 3)
    plt.imshow(flip_vh)
    plt.title("Pencerminan Keduanya (Flip Code = -1)")
    plt.show()

    img_draw = img_rgb.copy()

    cv.rectangle(img_draw, (int(w*0.7), int(h*0.1)), (int(w*0.95), int(h*0.8)), (0, 255, 0), 2)
    cv.circle(img_draw, (int(w*0.825), int(h*0.45)), 40, (255, 0, 0), 2)
    cv.putText(img_draw, "Angga Saputra Ramadhan", (int(w*0.05), int(h*0.92)),
               cv.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2, cv.LINE_AA)

    plt.figure(figsize=(6, 6))
    plt.imshow(img_draw)
    plt.title("Penggambaran Geometri & Teks")
    plt.show()
```

```python
import cv2 as cv
import numpy as np
import os

file_path = 'ktp.jpg'

if not os.path.exists(file_path):
    print(f"Error: File '{file_path}' tidak ditemukan.")
    print("Silakan upload file 'ktp.jpg' terlebih dahulu ke panel Files di Google Colab.")
else:
    img = cv.imread(file_path)
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

    mean_awal = img_rgb.mean()

    h, w = img_rgb.shape[:2]
    img_small = cv.resize(img_rgb, (w // 2, h // 2), interpolation=cv.INTER_LINEAR)
    img_restored = cv.resize(img_small, (w, h), interpolation=cv.INTER_LINEAR)

    mean_akhir = img_restored.mean()

    print(f"Rata-rata intensitas awal   : {mean_awal:.4f}")
    print(f"Rata-rata intensitas akhir  : {mean_akhir:.4f}")
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
import os

file_path = 'ktp.jpg'

if not os.path.exists(file_path):
    print(f"Error: File '{file_path}' tidak ditemukan.")
    print("Silakan upload file 'ktp.jpg' terlebih dahulu ke panel Files di Google Colab.")
else:
    def tambahkan_label_adaptif(image_path, nama_kegiatan):
        img = cv.imread(image_path)
        img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
        h, w, _ = img.shape

        bar_height = int(h * 0.1)

        overlay = img.copy()
        cv.rectangle(overlay, (0, h - bar_height), (w, h), (0, 0, 0), -1)

        alpha = 0.6
        img_result = cv.addWeighted(overlay, alpha, img, 1 - alpha, 0)

        font_scale = max(0.4, w / 1000.0)
        thickness = max(1, int(font_scale * 2))

        cv.putText(img_result, nama_kegiatan, (int(w * 0.03), h - int(bar_height * 0.35)),
                   cv.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

        return img_result

    output_kasus1 = tambahkan_label_adaptif(file_path, 'DOKUMENTASI PRAKTIKUM PCVK 2026 - POLINEMA')
    plt.figure(figsize=(8, 6))
    plt.imshow(output_kasus1)
    plt.title("Studi Kasus 1: Label Identitas Adaptif")
    plt.show()
```

```python
def buat_thumbnail_persegi(image_path, nama_toko):
    img = cv.imread(image_path)
    if img is None:
        raise FileNotFoundError(f"File '{image_path}' tidak ditemukan. Pastikan file sudah di-upload ke Google Colab.")
    img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    h, w, _ = img.shape

    sisi = max(h, w)

    kanvas = np.full((sisi, sisi, 3), 255, dtype=np.uint8)

    y_off = (sisi - h) // 2
    x_off = (sisi - w) // 2

    kanvas[y_off:y_off+h, x_off:x_off+w] = img

    font_scale = sisi / 800.0
    thickness = max(1, int(font_scale * 2))
    text_size = cv.getTextSize(nama_toko, cv.FONT_HERSHEY_DUPLEX, font_scale, thickness)[0]

    org_x = sisi - text_size[0] - int(sisi * 0.05)
    org_y = sisi - int(sisi * 0.05)

    cv.putText(kanvas, nama_toko, (org_x, org_y), cv.FONT_HERSHEY_DUPLEX, font_scale, (200, 200, 200), thickness + 2, cv.LINE_AA)
    cv.putText(kanvas, nama_toko, (org_x, org_y), cv.FONT_HERSHEY_DUPLEX, font_scale, (0, 0, 0), thickness, cv.LINE_AA)

    return kanvas
```
