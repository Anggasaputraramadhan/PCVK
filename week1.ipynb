```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
from skimage import io

url = "https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg"
image = io.imread(url)
image = cv2.resize(image, (0,0), fx=0.5, fy=0.5)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
tinggi, lebar, _ = image_rgb.shape

img_tugas_d2 = image_rgb.copy()
for y in range(lebar):
    img_tugas_d2[int(tinggi/2), y] = [255, 255, 255]
for x in range(tinggi):
    img_tugas_d2[x, int(lebar/2)] = [255, 255, 255]
    img_tugas_d2[x, x] = [255, 255, 255]

print("--- Hasil Modifikasi Tugas D2 ---")
cv2_imshow(cv2.cvtColor(img_tugas_d2, cv2.COLOR_RGB2BGR))

img_rb = image_rgb.copy()
img_rb[:, :, 1] = 0

img_gb = image_rgb.copy()
img_gb[:, :, 0] = 0

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(img_rb); plt.title("Kanal Red-Blue")
plt.subplot(1, 2, 2); plt.imshow(img_gb); plt.title("Kanal Green-Blue")
plt.show()

cropped_img = image_rgb[20:115, 25:120]
plt.figure(figsize=(4, 4))
plt.imshow(cropped_img)
plt.title("Potongan Baris 20-115, Kolom 25-120")
plt.show()

red_channel_slice = image_rgb[5:30, :, 0]
plt.figure(figsize=(6, 2))
plt.imshow(red_channel_slice, cmap='gray')
plt.title("Baris 5-30, Semua Kolom (Kanal Red)")
plt.show()

canvas = np.zeros((512, 512, 3), dtype=np.uint8)
np.random.seed(42)
for i in range(5):
    pt1 = (np.random.randint(10, 200), np.random.randint(10, 200))
    pt2 = (pt1[0] + np.random.randint(50, 150), pt1[1] + np.random.randint(50, 150))
    color = (int(np.random.randint(50, 255)), int(np.random.randint(50, 255)), int(np.random.randint(50, 255)))
    cv2.rectangle(canvas, pt1, pt2, color, thickness=-1)

plt.figure(figsize=(5, 5))
plt.imshow(canvas)
plt.title("5 Kotak Warna Acak")
plt.show()

from google.colab import drive
drive.mount('/content/drive')

img_ktm = cv2.imread("/content/drive/MyDrive/PCVK/KTM lama.jpg")

if img_ktm is not None:
    img_ktm_rgb = cv2.cvtColor(img_ktm, cv2.COLOR_BGR2RGB)
    img_ktm_rgb[200:350, 150:500] = [0, 0, 255]
    cv2.putText(img_ktm_rgb, "CONFIDENTIAL", (160, 280),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2, cv2.LINE_AA)

    plt.figure(figsize=(8, 6))
    plt.imshow(img_ktm_rgb)
    plt.title("Hasil Modifikasi KTM")
    plt.axis('on')
    plt.show()
else:
    print("File gambar tidak ditemukan, periksa kembali path direktori Anda.")
```
